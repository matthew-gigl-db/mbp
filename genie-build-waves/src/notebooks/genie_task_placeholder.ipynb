# Genie Task Placeholder

The `setup_build_waves_job` notebook swaps every task that points at this notebook for a `genie_task` (using `automation` to pick the `configuration_id`) via `jobs/reset`. Every `bundle deploy` puts the placeholders back, so re-run the setup job after each deploy.

If this notebook actually runs, the job was not patched, and it fails loudly so no wave runs without its Genie Code session.
